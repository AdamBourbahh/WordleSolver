## ¿Qué es la "información"?
Antes de explicar cualquier cosa, hay que introducir un concepto, la "información". 
Imaginemos que te digo un dato, cualquiera, hay alguna forma de medir cuanta "información" real te ha dado ese dato. Claude Shannon (por cierto, la IA Claude se llama así en su honor) razonó que la información esta inversamente relacionada con la probabilidad del evento, es decir:

* Si la probabilidad es del 100% ( $ p = 1 $), como "hoy ha salido el sol", la sorpresa es 0, y la información aportada es nula.
* Si la probabilidad es del 0% ( $ p = 0 $), como "Antonio ha muerto", la sorpresa es enorme, y la información aportada también.

Shannon buscaba una función matemática $I(p)$ que representara esta cantidad de información, y además debía cumplir una regla, *la aditividad de eventos independientes*. Es decir, si te dan dos datos distintos, la información total debe ser la suma de estos dos:  $$I(p_1 \cdot p_2) = I(p_1) + I(p_2)$$

En matemáticas, la única función que transforma el producto en suma es el logaritmo, y además, como debe ser decreciente, Shannon escogió el logaritmo negativo: $$I(x) = -\log_2(p(x))=\log_2\left(\frac{1}{p(x)}\right) $$

Al usar base 2, llamo a la unidad *bit* (binary digit), que es exactamente la cantidad de información necesaria para reducir la incertidumbre a la mitad

## La Entropía
La función anterior $I(x)$ funciona cuando el evento ya ha ocurrido, pero en Wordle, o cualquier sistema predictivo, introducimos una palabra y no sabemos qué colores vamos a recibir. Estos colores que recibimos son una variable aleatoria discreta (llamémosla $X$), como no sabemos que patrón de colores puede ocurrir, no se puede medir la información de un evento aislado. Por lo que lo que hacemos es calcular la Esperanza Matemática de toda la información que podríamos recibir, a este valor lo llamaremos *Entropía ($H$)*.

Dado que la Entropía no es más que la Esperanza, la fórmula de esta es: $$H(X) = \mathbb{E}[I(X)]=\sum_{x \in X} p(x) \log_2\left(\frac{1}{p(x)}\right) $$

## Maximizar la Entropía.

Hemos dicho que la Entropía es el valor medio de información que podemos esperar, por lo que lo lógico sería, maximizar esta, que sucede cuando todos los eventos son equiprobables. El equivalente en el wordle sería el siguiente:
Imaginemos que nos quedan 100 palabras posibles, y estamos entre dos elecciones, Palabra A o Palabra B:

* **Palabra A (Baja entropía)**: Imaginemos que es KIWIS, como muy pocas palabras tienes la K y la W, la probabilidad de que nos devuelva todo gris, es altísima, y además, al ser letras poco comunes, no nos da mucha información, eso es intuitivamente que tenga poca entropía.
* **Palabra B (Alta Entropía)**: Imaginemos que es TARES, dado que tanto la T,R,S y las vocales A y E son muy comunes, la información que recibimos con todo gris, una verde, amarilla..., es bastante, pues reducimos el espacio de búsqueda al descartar palabras con dichas letras, eso es tener alta entropía.

Por tanto, maximizar la entropía no es más que buscar la palabra que fragmenta la lista de candidatos en trozos pequeños y equilibrados.
