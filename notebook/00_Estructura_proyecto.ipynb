# Wordle Solver : Optimización Mediante Teoría de la Información
Este proyecto es la construcción de un programa capaz de resolver el juego Wordle, de forma eficiente. 
## Introducción y Descripción del Problema
Wordle es un juego de deducción lingüistica cuyo objetivo es adivinar una palabra secreta de 5 letras en como mucho 6 intentos. Cada intento nos dice si en cada posición hemos acertado o no la letra, y si cierta letra existe, pero en otra posición. Si la palabra fuese FENCE, y pusiésemos RENFE, nos diria: 

* La `R` no está
* `E`, `N` y `E` están correctas
* `F` está en el lugar incorrecto pero está.

La lógica humana a la hora de jugar a Wordle suele ser siempre la misma, probar las palabras que nos quiten la mayor cantidad de opciones posibles, esto precisamente se puede lograr a través de la Entropía de Shannon ($H$), por ello he elegido este enfoque.

## Arquitectura del Software
El sistema está diseñado en C++ bajo principios de diseño orientado a objetos y separación de responsabilidades:
* Módulo de Datos (**Conjunto**): Encargado de la persistencia y lectura eficiente en disco de los diccionarios.
* Módulo de Reglas (**Logica**): Actúa como el motor del juego. 
* Módulo de Decisión (**Entropia**): Implementa el cálculo matricial/frecuencial de la información esperada y el filtrado iterativo del conjunto de candidatos.
